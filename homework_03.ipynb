```python
def transpose_matrix(matrix):
    result = []
    for i in range(len(matrix[0])):
        new_row = []
        for j in range(len(matrix)):
            new_row.append(matrix[j][i])
            result.append(new_row)
            return result
print(transpose_matrix([[1,2,3],[4,5,6]]))
```

```python
def unique_common_elements(list1, list2):
    common = []
    for item in list1:
        if item in list2 and item  not in common:
            common.append(item)
    return sorted(common)

print(unique_common_elements([1,2,2,3],[3,4,2,2]))
```

```python
def word_frequency(text):
    text = text.replace('.','').replace('.','').replace('!','').replace('?','')
    words = text.lower().split()
    freq = {}
    for word in words:
        if word in freq:
            freq[word] += 1
        else:
            freq[word] = 1
    return freq
print(word_frequency("Hello,world! Hello again."))    
```

```python
def group_by_score(students):
    groups = {}
    for name, score in students:
        if score not in groups:
            groups[score] = []
        groups[score].append(name)
    for score in groups:
        groups[score].sort()
    return groups

print(group_by_score([('Alice', 85), ('Bob', 92), ('Charlie', 85)]))
```

```python
def merge_intervals(intervals):
    if not intervals:
        return []
    merged = [intervals[0]]
    for current in intervals[1:]:
        last = merged[-1]
        if current[0] <= last[1]:
            merged[-1] = (last[0], max(last[1], current[1]))
        else:
            merged.append(current)
    return merged

print(merge_intervals([(1, 3), (2, 6), (8, 10), (15, 18)]))
```

```python
def access_diff(yesterday, today):
    only_yesterday = []
    only_today = []
    for user in yesterday:
        if user not in today:
            only_yesterday.append(user)
    for user in today:
        if user not in yesterday:
            only_today.append(user)
    only_yesterday.sort()
    only_today.sort()
    return (only_yesterday, only_today)

print(access_diff(['Alice', 'Bob', 'Charlie'], ['Bob', 'David']))
```

```python
def invert_dictionary(d):
    inverted = {}
    for key, value in d.items():
        if value not in inverted:
            inverted[value] = []
        inverted[value].append(key)
    for value in inverted:
        inverted[value].sort()
    return inverted

print(invert_dictionary({'a': 1, 'b': 2, 'c': 1}))
```

```python
def flatten_list(nested_list):
    result = []
    for item in nested_list:
        if type(item) == list:
            result.extend(flatten_list(item))
        else:
            result.append(item)
    return result

print(flatten_list([1, [2, [3, 4], 5], 6]))
```

```python
def create_contact_book(contacts):
    phone_book = {}
    for name, number in contacts:
        phone_book[number] = name
    return phone_book

print(create_contact_book([('Alice', '123'), ('Bob', '456'), ('Charlie', '789')]))
```

```python
def analyze_logs(logs):
    ip_pages = {}
    for log in logs:
        parts = log.split()
        ip = parts[0]
        page = parts[6]
        if ip not in ip_pages:
            ip_pages[ip] = set()
        ip_pages[ip].add(page)
    return ip_pages

sample_logs = [
    '8.8.8.8 - - [11/Oct/2025:14:00:01 -0700] "GET /home HTTP/1.1"',
    '192.168.1.101 - - [11/Oct/2025:14:00:02 -0700] "GET /login HTTP/1.1"',
    '8.8.8.8 - - [11/Oct/2025:14:00:03 -0700] "GET /products HTTP/1.1"'
]
print(analyze_logs(sample_logs))
```
