```python
class User:
    def __init__(self, username, password):
        self.username = username
        self._password = None  # Initialize as None first
        self.set_password(password)  # Use set_password to validate
    
    def set_password(self, new_password):
        if len(new_password) < 8:
            raise ValueError("Password must be at least 8 characters long")
        self._password = new_password
    
    def check_password(self, password_to_check):
        return self._password == password_to_check

# Test the User class
if __name__ == "__main__":
    try:
        # Test valid password
        user1 = User("john_doe", "password123")
        print(f"User '{user1.username}' created successfully!")
        
        # Test password checking
        print("Check correct password:", user1.check_password("password123"))
        print("Check wrong password:", user1.check_password("wrongpass"))
        
        # Test password change
        user1.set_password("newpassword456")
        print("After password change:")
        print("Check old password:", user1.check_password("password123"))
        print("Check new password:", user1.check_password("newpassword456"))
        
        # Test invalid password (should raise ValueError)
        print("\nTesting invalid password...")
        user2 = User("test_user", "short")
        
    except ValueError as e:
        print(f"Error: {e}")
    
    # Demonstrate encapsulation - no direct access to password
    print(f"\nUsername (public): {user1.username}")
    # print(user1._password)  # This would work but is against encapsulation principle
    # There's no get_password() method as required
```

```python
class StringUtils:
    """
    A utility class with static methods for working with strings.
    """
    
    @staticmethod
    def reverse_string(s):
        """
        Takes a string s and returns it reversed.
        """
        return s[::-1]
    
    @staticmethod
    def is_palindrome(s):
        """
        Takes a string s and returns True if it is a palindrome, False otherwise.
        """
        # Convert to lowercase and remove non-alphanumeric characters for more accurate palindrome check
        cleaned = ''.join(char.lower() for char in s if char.isalnum())
        return cleaned == cleaned[::-1]
    
    @staticmethod
    def count_vowels(s):
        """
        Takes a string s and returns the number of vowels (a, e, i, o, u) in it.
        Counts both lowercase and uppercase vowels.
        """
        vowels = "aeiouAEIOU"
        count = 0
        for char in s:
            if char in vowels:
                count += 1
        return count


# Test the StringUtils class
if __name__ == "__main__":
    # Test reverse_string
    print("Reverse 'hello':", StringUtils.reverse_string("hello"))
    print("Reverse 'Python':", StringUtils.reverse_string("Python"))
    
    # Test is_palindrome
    print("\nIs 'racecar' a palindrome?:", StringUtils.is_palindrome("racecar"))
    print("Is 'A man a plan a canal Panama' a palindrome?:", StringUtils.is_palindrome("A man a plan a canal Panama"))
    print("Is 'hello' a palindrome?:", StringUtils.is_palindrome("hello"))
    
    # Test count_vowels
    print("\nVowels in 'Hello World':", StringUtils.count_vowels("Hello World"))
    print("Vowels in 'Python Programming':", StringUtils.count_vowels("Python Programming"))
    print("Vowels in 'AEIOUaeiou':", StringUtils.count_vowels("AEIOUaeiou"))
```

```python
class Person:
    """
    A class that stores a person's age with validation.
    """
    
    def __init__(self, name, age):
        """
        Initialize a Person with name and age.
        """
        self._name = name
        self._age = age  # This will use the setter for validation
    
    @property
    def age(self):
        """
        Getter for age - returns the value of _age.
        """
        return self._age
    
    @age.setter
    def age(self, value):
        """
        Setter for age - validates that age is between 0 and 120.
        """
        if not (0 <= value <= 120):
            raise ValueError("Age must be between 0 and 120")
        self._age = value
    
    @property
    def is_adult(self):
        """
        Read-only property that returns True if age is 18 or greater.
        """
        return self._age >= 18
    
    @property
    def name(self):
        """
        Getter for name.
        """
        return self._name
    
    @name.setter
    def name(self, value):
        """
        Setter for name.
        """
        self._name = value


# Test the Person class
if __name__ == "__main__":
    # Create a person with valid age
    person1 = Person("Alice", 25)
    print(f"{person1.name} is {person1.age} years old")
    print(f"Is {person1.name} an adult? {person1.is_adult}")
    
    # Test age validation
    try:
        person2 = Person("Bob", 150)  # This should raise ValueError
    except ValueError as e:
        print(f"\nError creating Bob: {e}")
    
    # Test age setter validation
    person3 = Person("Charlie", 30)
    print(f"\n{person3.name} is {person3.age} years old")
    
    try:
        person3.age = -5  # This should raise ValueError
    except ValueError as e:
        print(f"Error setting age: {e}")
    
    try:
        person3.age = 130  # This should raise ValueError
    except ValueError as e:
        print(f"Error setting age: {e}")
    
    # Test valid age change
    person3.age = 35
    print(f"After change: {person3.name} is now {person3.age} years old")
    
    # Test is_adult property
    child = Person("David", 10)
    adult = Person("Eve", 18)
    
    print(f"\n{child.name} (age {child.age}) is adult: {child.is_adult}")
    print(f"{adult.name} (age {adult.age}) is adult: {adult.is_adult}")
    
    # Test that is_adult is read-only
    print(f"\nTrying to modify is_adult property...")
    try:
        adult.is_adult = False  # This should fail
    except AttributeError as e:
        print(f"Cannot modify is_adult: {e}")
```

```python
class Vector:
    def init(self, x, y):
        self.x = x
        self.y = y
    
    def add(self, other):
        result = Vector()
        result.init(self.x + other.x, self.y + other.y)
        return result
    
    def sub(self, other):
        result = Vector()
        result.init(self.x - other.x, self.y - other.y)
        return result
    
    def mul(self, scalar):
        result = Vector()
        result.init(self.x * scalar, self.y * scalar)
        return result
    
    def str(self):
        return f"Vector({self.x}, {self.y})"
    
    def repr(self):
        return f"Vector({self.x}, {self.y})"

# Test the Vector class
print("=== Vector Class Demo ===\n")

# Create vectors
v1 = Vector()
v1.init(2, 3)

v2 = Vector()
v2.init(5, 7)

print("Vector 1:", v1.str())
print("Vector 2:", v2.str())

# Test addition
v3 = v1.add(v2)
print(f"\nAddition: {v1.str()} + {v2.str()} = {v3.str()}")

# Test subtraction
v4 = v1.sub(v2)
print(f"Subtraction: {v1.str()} - {v2.str()} = {v4.str()}")

# Test scalar multiplication
v5 = v1.mul(3)
print(f"Scalar multiplication: {v1.str()} * 3 = {v5.str()}")

v6 = v2.mul(2)
print(f"Scalar multiplication: {v2.str()} * 2 = {v6.str()}")

# Test multiple operations
print(f"\n--- Multiple Operations ---")
result = v1.add(v2).mul(2)
print(f"({v1.str()} + {v2.str()}) * 2 = {result.str()}")

# Create more vectors
v7 = Vector()
v7.init(1, 1)

v8 = Vector()
v8.init(4, 2)

print(f"\n--- More Examples ---")
print(f"Vector A: {v7.str()}")
print(f"Vector B: {v8.str()}")
print(f"A + B = {v7.add(v8).str()}")
print(f"B - A = {v8.sub(v7).str()}")
print(f"A * 5 = {v7.mul(5).str()}")
```

```python
import json

class Task:
    # simple constructor
    def create(self, description, due_date, status="pending"):
        self.description = description
        self.due_date = due_date
        self.status = status

class TaskManager:
    def init(self):
        self.tasks = []

    # Add a new task
    def add_task(self, description, due_date, status="pending"):
        task = Task()
        task.create(description, due_date, status)
        self.tasks.append(task)
        print("Task added.")

    # View all tasks
    def view_tasks(self):
        if len(self.tasks) == 0:
            print("No tasks found.")
        else:
            for i, t in enumerate(self.tasks, 1):
                print("ID:", i)
                print(" Description:", t.description)
                print(" Due Date:", t.due_date)
                print(" Status:", t.status)
                print("-" * 25)

    # Change task status by ID
    def change_status_by_id(self, task_id, new_status):
        index = task_id - 1
        if 0 <= index < len(self.tasks):
            self.tasks[index].status = new_status
            print("Status updated.")
        else:
            print("Invalid ID.")

    # Change task status by description
    def change_status_by_description(self, description, new_status):
        found = False
        for t in self.tasks:
            if t.description == description:
                t.status = new_status
                found = True
        if found:
            print("Status updated.")
        else:
            print("Task not found.")

    # Delete task
    def delete_task(self, task_id):
        index = task_id - 1
        if 0 <= index < len(self.tasks):
            del self.tasks[index]
            print("Task deleted.")
        else:
            print("Invalid ID.")

    # Save tasks to file
    def save_to_file(self, filename):
        data = []
        for t in self.tasks:
            data.append({
                "description": t.description,
                "due_date": t.due_date,
                "status": t.status
            })

        with open(filename, "w") as f:
            json.dump(data, f, indent=4)
        print("Tasks saved.")

    # Load tasks from file
    def load_from_file(self, filename):
        try:
            with open(filename, "r") as f:
                data = json.load(f)

            self.tasks = []
            for item in data:
                t = Task()
                t.create(item["description"], item["due_date"], item["status"])
                self.tasks.append(t)

            print("Tasks loaded.")
        except:
            print("Could not load file.")


def main():
    tm = TaskManager()

    while True:
        print("\nTask Manager")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Change Status by ID")
        print("4. Change Status by Description")
        print("5. Delete Task")
        print("6. Save Tasks")
        print("7. Load Tasks")
        print("0. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            d = input("Description: ")
            due = input("Due date: ")
            s = input("Status (default pending): ") or "pending"
            tm.add_task(d, due, s)

        elif choice == "2":
            tm.view_tasks()

        elif choice == "3":
            tid = int(input("Task ID: "))
            ns = input("New status: ")
            tm.change_status_by_id(tid, ns)

        elif choice == "4":
            desc = input("Description: ")
            ns = input("New status: ")
            tm.change_status_by_description(desc, ns)

        elif choice == "5":
            tid = int(input("Task ID: "))
            tm.delete_task(tid)

        elif choice == "6":
            tm.save_to_file("tasks.json")

        elif choice == "7":
            tm.load_from_file("tasks.json")

        elif choice == "0":
            break

        else:
            print("Invalid choice!")

if __name__== "__main__":
    main()
```

```python
# Abstract Document class
class Document:
    def display(self):
        pass


# Concrete PDF document
class PDFDocument(Document):
    def display(self):
        print("Displaying PDF document")


# Concrete Word document
class WordDocument(Document):
    def display(self):
        print("Displaying Word document")


# Concrete Web document
class WebDocument(Document):
    def display(self):
        print("Displaying Web document")


# Factory class
class DocumentFactory:
    @staticmethod
    def create_document(doc_type):
        if doc_type == "pdf":
            return PDFDocument()
        elif doc_type == "word":
            return WordDocument()
        elif doc_type == "web":
            return WebDocument()
        else:
            print("Unknown document type.")
            return None


# Client code
def main():
    print("Document Factory Example")
    print("pdf / word / web")

    doc_type = input("Enter document type: ")

    doc = DocumentFactory.create_document(doc_type)

    if doc:
        doc.display()


main()
```

# Document — базовый класс (родитель), содержит метод display()
# PDFDocument/WordDocument/WebDocument — конкретные подклассы (override display)
# Полиморфизм: doc.display() работает по-разному для разных типов документов
# Factory — DocumentFactory.create_document выбирает нужный класс по строке
# create_document — реализация паттерна Factory (создание объектов по запросу)
# Статический метод: вызов без объекта, удобно для фабрик
# Неизвестный тип → возвращает None (ошибка обработки минимальная)
# Client code не знает, какой класс создаётся — только вызывает factory
# Это разделяет создание объекта и его использование (principle “separation”)
# display() в базовом классе пустой — имитирует “абстрактный” метод
