```python
# Объявляем декоратор print_boundary
# Этот декоратор выводит строку "--- START ---" перед вызовом функции
# и "--- END ---" после её выполнения
def print_boundary(func):

    # Функция-обёртка, которая заменяет собой оригинальную функцию
    # *args и **kwargs позволяют передавать любые аргументы
    def wrapper(*args, **kwargs):
        print("--- START ---")   # Печатаем начало перед выполнением функции

        result = func(*args, **kwargs)   # Вызываем оригинальную функцию

        print("--- END ---")     # Печатаем конец после выполнения функции

        return result            # Возвращаем результат оригинальной функции
    return wrapper               # Возвращаем функцию-обёртку


# -----------------------------
# Используем декоратор для функции ниже
# -----------------------------
@print_boundary
def hello():
    print("Hello world!")  # Эта функция будет автоматически окружена START/END


# Запуск программы — вызываем функцию hello
hello()
```

```python
# Декоратор uppercase_result
# Этот декоратор принимает функцию, которая должна возвращать строку,
# и преобразует результат в верхний регистр (UPPERCASE).
def uppercase_result(func):

    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)  # вызываем оригинальную функцию

        # Проверяем, что результат — строка
        if not isinstance(result, str):
            raise TypeError("Функция должна возвращать строку!")  # ошибка, если не строка

        return result.upper()  # возвращаем строку в верхнем регистре

    return wrapper


# -----------------------------
# Пример использования декоратора
# -----------------------------
@uppercase_result
def greet():
    return "hello world"


print(greet())   # Результат: HELLO WORLD
```

```python
# Функция-генератор countdown(n)
# Она должна выдавать (yield) числа от n до 0 включительно.
def countdown(n):

    # Цикл от n до 0, включая 0
    while n >= 0:
        yield n   # возвращаем текущее значение через yield
        n -= 1    # уменьшаем число на 1
        

# -----------------------------
# Пример использования генератора
# -----------------------------
for num in countdown(5):
    print(num)
```

```python
# Класс-итератор ReverseIterator
# Он принимает строку и проходит по её символам в обратном порядке.
class ReverseIterator:

    # Конструктор принимает строку
    def __init__(self, text):
        self.text = text              # сохраняем строку
        self.index = len(text) - 1    # начинаем с последнего символа

    # Метод iter должен вернуть сам итератор
    def __iter__(self):
        return self

    # Метод next возвращает следующий элемент (в обратном порядке)
    def __next__(self):
        # Если индекс вышел за пределы — итерация закончилась
        if self.index < 0:
            raise StopIteration

        # Берём текущий символ
        char = self.text[self.index]

        # Передвигаем индекс влево (к началу строки)
        self.index -= 1

        return char


# -----------------------------
# Пример использования итератора
# -----------------------------
it = ReverseIterator("abc")

for ch in it:
    print(ch)
```

```python
# Декоратор log_call
# Он выводит:
# - имя вызываемой функции
# - позиционные аргументы
# - именованные аргументы
# - возвращаемое значение функции
def log_call(func):

    def wrapper(*args, **kwargs):
        # Печатаем имя функции
        print(f"Calling function: {func.__name__}")

        # Печатаем позиционные аргументы
        print(f"Positional args: {args}")

        # Печатаем именованные аргументы
        print(f"Keyword args: {kwargs}")

        # Вызываем оригинальную функцию
        result = func(*args, **kwargs)

        # Печатаем возвращаемое значение
        print(f"Return value: {result}")

        return result  # Возвращаем результат

    return wrapper


# -----------------------------
# Пример использования декоратора
# -----------------------------
@log_call
def add(a, b):
    return a + b


result = add(2, 3)
```

```python
# Параметризованный декоратор repeat(n)
# Он вызывает декорированную функцию n раз
# и возвращает список результатов всех вызовов.

def repeat(n):

    # Это настоящий декоратор, который принимает функцию
    def decorator(func):

        # Функция-обёртка, которая будет вызываться вместо оригинальной функции
        def wrapper(*args, **kwargs):
            results = []   # список для хранения результатов

            # Выполняем функцию n раз
            for _ in range(n):
                results.append(func(*args, **kwargs))

            return results  # возвращаем список всех результатов

        return wrapper  # возвращаем обёртку

    return decorator  # возвращаем декоратор


# -----------------------------
# Пример использования
# -----------------------------
@repeat(3)
def square(x):
    return x * x


print(square(2))   # Ожидаемый результат: [4, 4, 4]
```

```python
# Функция-генератор file_lines_generator
# Она принимает путь к файлу и построчно (по одной строке)
# выдаёт строки этого файла.
# Если файл не найден — обрабатываем исключение FileNotFoundError.

def file_lines_generator(filepath):

    try:
        # Открываем файл в режиме чтения
        with open(filepath, "r", encoding="utf-8") as f:

            # Идём по файлу строка за строкой
            for line in f:
                yield line   # выдаём строку генератором

    except FileNotFoundError:
        print(f"File not found: {filepath}")   # сообщение об ошибке
        return  # завершаем работу генератора
        

# -----------------------------
# Пример использования
# -----------------------------
for line in file_lines_generator("example.txt"):
    print(line, end="")   # end="" чтобы не дублировать перенос строки
```

```python
# Генератор prime_generator(limit)
# Он должен выдавать (yield) все простые числа
# от 2 до заданного предела limit включительно.

def prime_generator(limit):

    # Функция для проверки: является ли число простым
    def is_prime(n):
        if n < 2:
            return False
        # Проверяем делители от 2 до корня числа
        for d in range(2, int(n**0.5) + 1):
            if n % d == 0:
                return False
        return True

    # Основной цикл генератора
    for num in range(2, limit + 1):
        if is_prime(num):      # если число простое
            yield num          # выдаём его


# -----------------------------
# Пример использования
# -----------------------------
for p in prime_generator(20):
    print(p)
```

```python
# Декоратор cache_result
# Он кэширует результаты работы функции.
# При первом вызове с определёнными аргументами — функция выполняется.
# При следующих вызовах с теми же аргументами — возвращается сохранённый результат.

def cache_result(func):
    cache = {}   # словарь для хранения результатов

    def wrapper(*args, **kwargs):
        key = (args, tuple(kwargs.items()))  # создаём ключ по аргументам

        if key in cache:
            return cache[key]   # возвращаем сохранённый результат

        result = func(*args, **kwargs)  # выполняем функцию впервые
        cache[key] = result             # сохраняем в кэш
        return result

    return wrapper


# -----------------------------
# Пример использования
# -----------------------------
@cache_result
def slow_add(a, b):
    print("Выполняем функцию...")  # покажет только при первом вызове
    return a + b


print(slow_add(2, 3))   # первый вызов — выполняется функция
print(slow_add(2, 3))   # второй вызов — берётся из кэша, функция не выполняется
```

```python
# Генератор nested_list_unpacker
# Он принимает вложенный список любой глубины
# и выдаёт (yield) все элементы по одному — в плоской форме.

def nested_list_unpacker(nested_list):

    for item in nested_list:
        if isinstance(item, list):
            # если элемент — список, рекурсивно распаковываем его
            yield from nested_list_unpacker(item)
        else:
            # если элемент не список — выдаём его напрямую
            yield item


# -----------------------------
# Пример использования
# -----------------------------
data = [1, [2, 3], [4, [5, 6]]]

for x in nested_list_unpacker(data):
    print(x)
```
