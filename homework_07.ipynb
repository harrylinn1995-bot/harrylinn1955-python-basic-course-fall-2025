```python
import numpy as np

# Create an array of 365 random temperatures between -10 and 35
temperatures = np.random.uniform(-10, 35, 365)

# 1. Average temperature for the whole year
year_avg = np.mean(temperatures)

# 2. Average temperature for summer (day 150 to 240)
summer_avg = np.mean(temperatures[149:240])

# 3. Standard deviation for the whole year
year_std = np.std(temperatures)

print("Average temperature (whole year):", year_avg)
print("Average summer temperature:", summer_avg)
print("Standard deviation (year):", year_std)
```

```python
import numpy as np

# Stock prices for the week
prices = np.array([150, 152, 149, 155, 157])

# 1. Buy on Monday and sell on Friday (profit)
profit = prices[-1] - prices[0]
print("Profit:", profit)

# 2. Buy 10 stocks every day (money spent per day)
money_spent = prices * 10
print("Money spent each day:", money_spent)
```

1️⃣ Array indexing
prices[0] = first day
prices[-1] = last day
2️⃣ Profit formula
profit = selling_price – buying_price 
3️⃣ NumPy multiplication
prices * 10 
→ element-wise multiplication
→ each day's cost

```python
from PIL import Image
import numpy as np
from IPython.display import display

# 1. Load your image (your image is inside images/images.jpg)
img = Image.open("images/images.jpg")

# Convert to numpy array
arr = np.array(img)

# 2. Negative image
negative = 255 - arr
negative_img = Image.fromarray(negative.astype(np.uint8))

# 3. Upside-down image
upside = arr[::-1]
upside_img = Image.fromarray(upside.astype(np.uint8))

# Display results
print("Original Image:")
display(img)

print("Negative Image:")
display(negative_img)

print("Upside-down Image:")
display(upside_img)
```

1️⃣ Image → array
arr = np.array(img)
➡️ ပုံကို pixel array အဖြစ်ပြောင်းတာ။
2️⃣ Negative image
negative = 255 - arr
➡️ Pixel value ကို ပြောင်းပြီး အရောင်ပြောင်းတာ။
3️⃣ Upside-down
upside = arr[::-1]
➡️ Row တွေ ပြောင်းလဲပြီး ပုံအောက်ခြေနဲ့အပေါ်ခြေ invert လုပ်တာ။

```python
import numpy as np

# Create a 5x7 sales matrix (random integers 1–100)
sales = np.random.randint(1, 101, (5, 7))
print("Sales Matrix:\n", sales)

# 1. Total revenue for the week
total_revenue = np.sum(sales)
print("\nTotal revenue for the week:", total_revenue)

# 2. Day with maximum total sales (column index)
day_max_sales = np.argmax(np.sum(sales, axis=0))
print("Day with max sales (column index):", day_max_sales)

# 3. Product with minimum total sales (row index)
product_worst = np.argmin(np.sum(sales, axis=1))
print("Worst selling product (row index):", product_worst)
```

```python
import numpy as np

# 5 students × 3 exams (random scores 50 to 100)
scores = np.random.randint(50, 101, (5, 3))
print("Original Scores:\n", scores)

# Bonus for each exam
bonus = np.array([5, 0, 10])   # exam1 +5, exam2 +0, exam3 +10

# Add bonus using broadcasting
final_scores = scores + bonus

print("\nFinal Scores with Bonus:\n", final_scores)
```

1️⃣ scores = 5×3 matrix
➡️ 5 student, 3 exam marks.
2️⃣ bonus = [5, 0, 10]
➡️ Exam 1 +5, Exam 2 +0, Exam 3 +10.
3️⃣ scores + bonus
➡️ Broadcasting ထောက်ပံ့ပြီး bonus ကို student အားလုံး에게 auto ပေါင်းပေးတယ်।
 ဘာလို broadcasting သုံးရလဲ?

```python
import numpy as np

# Log file as strings
logs = np.array(["200", "404", "200", "500", "200", "404"])

# 1. Convert to integers
logs_int = logs.astype(int)
print("Integer log codes:", logs_int)

# 2. Calculate percentage of successful requests (code 200)
success_count = np.sum(logs_int == 200)
total = len(logs_int)
percentage = (success_count / total) * 100

print("Successful request percentage:", percentage, "%")
```

```python
import numpy as np

ratings = np.array([
    [5, 4, 0, 1],
    [1, 0, 5, 4],
    [5, 4, 1, 0],
    [0, 5, 4, 5],
    [1, 1, 5, 4]
])

diff = np.abs(ratings - ratings[0])
total_diff = np.sum(diff, axis=1)

# Exclude user 0 by looking from index 1 onward
similar_user = np.argmin(total_diff[1:]) + 1

print("Most similar user:", similar_user)
```
