```python
def word_counter(file_path):
    """Count lines, words, and non-space characters from a text file."""
    try:
        with open(file_path, 'r', encoding='utf-8') as file:   # File ကို UTF-8 နဲ့ဖွင့်တယ်
            lines = file.readlines()                          # Line list အဖြစ် ဖတ်တယ်

            line_count = len(lines)                           # Line အရေအတွက်
            word_count = 0                                    # Word counter 초기화
            char_count = 0                                    # Character counter 초기화 (space မပါ)

            for line in lines:
                words = line.split()                          # Whitespace ဖြင့် words ခွဲထားတယ်
                word_count += len(words)                      # Word အရေအတွက် စုတယ်

                char_count += len(line.replace(' ', ''))      # Space ဖယ်ပြီး character တွေတွက်တယ်

            return {
                'lines': line_count,                          # Line count result
                'words': word_count,                          # Word count result
                'chars': char_count                           # Char count result
            }

    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")        # File မတွေ့လျင် error ပြတယ်
        return None
    except Exception as e:
        print(f"Error reading file: {e}")                     # အခြား error
        return None


# Example usage
result = word_counter('data/sample_text.txt')                 # Function ကို file path နဲ့ခေါ်တယ်
if result:                                                    # Result ရှိမှ print
    print(result)
```

```python
from pathlib import Path

def save_list_to_file(string_list, file_path):
    """
    Save each item from a string list as a new line in a text file.
    """
    try:
        path = Path(file_path)                         # File path ကို Path object ဖြစ်အောင်ပြောင်း
        
        path.parent.mkdir(parents=True, exist_ok=True) # Folder မရှိသေးရင် create လုပ်
        
        with open(path, 'w', encoding='utf-8') as file:  # UTF-8 နဲ့ file ဖြင့်ရေး
            for string in string_list:
                file.write(string + '\n')              # Element တစ်ခုချင်းကို line တစ်ကြောင်းစီအဖြစ် သိမ်း
        
        print(f"Successfully saved {len(string_list)} lines to: {file_path}")
        return True                                    # Success
        
    except Exception as e:
        print(f"Error saving file: {e}")               # Error handling
        return False


# Example usage
if __name__ == "__main__":                             # Main program check (ပြင်ပ run 时)
    my_list = [
        "Hello, World!",
        "Python programming",
        "File handling example",
        "နေကောင်းလား?",    # Unicode OK
        "最后一行"          # Unicode OK
    ]

    success = save_list_to_file(my_list, "output/my_files/data.txt")  # File save call
    
    if success:
        print("File saved successfully!")              # Confirmation message
```

```python
def search_in_file(file_path, search_word):
    """
    Search for lines that contain a specific word (case-insensitive).
    """
    try:
        found_lines = []                                # Result list

        with open(file_path, 'r', encoding='utf-8') as file:  # File open
            for line in file:
                if search_word.lower() in line.lower():        # Case-insensitive compare
                    found_lines.append(line.strip())           # Clean line add

        return found_lines                              # Return found lines

    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")  # File not found
        return []
    except Exception as e:
        print(f"Error reading file: {e}")               # Other errors
        return []


# Example usage
if __name__ == "__main__":                              # Main program check
    search_words = ["python", "данных", "синтаксис", "programming"]

    for word in search_words:
        print(f"\nSearching for '{word}':")
        results = search_in_file('data/sample_text.txt', word)

        if results:
            print(f"Found {len(results)} line(s):")     # Count lines found
            for i, line in enumerate(results, 1):       # Print each matched line
                print(f"{i}. {line}")
        else:
            print("No matches found.")                   # If nothing found
```

```python
import csv

def read_csv_to_dict(file_path):
    """
    Read a CSV file and return list of rows as dictionaries.
    """
    try:
        with open(file_path, 'r', encoding='utf-8') as file:   # CSV file ကို UTF-8 နဲ့ဖွင့်
            csv_reader = csv.DictReader(file)                 # First row → headers, each row → dict

            data = list(csv_reader)                           # DictReader object ကို list ပြောင်း
            return data                                       # List of dicts return

    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")        # File မရှိလျင်
        return []

    except Exception as e:
        print(f"Error reading CSV file: {e}")                 # အခြား CSV error
        return []


# Example usage
if __name__ == "__main__":                                    # Main program check
    users = read_csv_to_dict('data/users.csv')                # CSV → dict list

    print("CSV Data as List of Dictionaries:")
    print("-" * 40)

    for i, user in enumerate(users, 1):                       # Row တစ်ခုစီ print
        print(f"Record {i}:")
        for key, value in user.items():                       # Key/value each column show
            print(f"  {key}: {value}")
        print()
```

```python
from pathlib import Path

def merge_files(file_paths, output_file):
    """
    Merge multiple text files into one output file (each with header).
    """
    merged_count = 0                                       # Merge success counter

    Path(output_file).parent.mkdir(parents=True, exist_ok=True)  
    # Output folder မရှိရင် create လုပ်

    with open(output_file, 'w', encoding='utf-8') as outfile:    # Output file open (overwrite mode)
        for i, file_path in enumerate(file_paths, 1):            # File list ကို index 1 થી loop
            try:
                with open(file_path, 'r', encoding='utf-8') as infile:  # Input file open
                    content = infile.read().strip()              # All content read + trim

                    outfile.write(f"=== File {i}: {Path(file_path).name} ===\n")  
                    # Header: file number + filename

                    outfile.write(content)                       # Actual content overwrite
                    outfile.write("\n\n")                        # Blank lines for separation

                    merged_count += 1                            # Success file increment

            except FileNotFoundError:
                continue                                          # Missing file → skip only

    return merged_count                                           # Return number of merged files


# Example usage
files = ['data/file1.txt', 'data/file2.txt', 'data/file3.txt']
result = merge_files(files, 'data/merged/combined.txt')
print(f"Merged {result} files")                  # Show final count
```

```python
import csv

def filter_csv(file_path, output_path, filter_func):
    """
    Filter CSV rows using a condition function and save result to new CSV.
    """
    filtered_count = 0                                   # Count rows that match filter

    with open(file_path, 'r', encoding='utf-8') as infile, \
         open(output_path, 'w', encoding='utf-8', newline='') as outfile:

        reader = csv.DictReader(infile)                  # Read CSV as dict rows
        writer = csv.DictWriter(outfile, fieldnames=reader.fieldnames)  
        # Writer uses same headers

        writer.writeheader()                             # Write header row first

        for row in reader:                               # Loop through each CSV row
            if filter_func(row):                         # Apply filtering condition
                writer.writerow(row)                     # Write matched row
                filtered_count += 1                      # Count matched rows

    return filtered_count                                # Return number of matches


# Test functions
def test_filter_csv():
    """Test the filter_csv function with different conditions"""

    count1 = filter_csv(
        'data/employees.csv', 'data/moscow_employees.csv',
        lambda row: row['city'] == 'Москва'              # Filter by city
    )
    print(f"Москва employees: {count1}")

    count2 = filter_csv(
        'data/employees.csv', 'data/senior_employees.csv',
        lambda row: int(row['age']) > 30                 # Filter by age > 30
    )
    print(f"Employees over 30: {count2}")

    count3 = filter_csv(
        'data/employees.csv', 'data/high_salary.csv',
        lambda row: int(row['salary']) >= 90000          # Filter by salary >= 90000
    )
    print(f"High salary employees: {count3}")


# Run tests
if __name__ == "__main__":                                # Main check
    test_filter_csv()
```

```python
import json

def apply_discount(input_file, min_price, discount_percent, output_file):
    """
    Apply discount to products whose price >= min_price,
    then save updated data to a new JSON file.
    """
    try:
        with open(input_file, 'r', encoding='utf-8') as f:   # Read input JSON
            products = json.load(f)                         # Load as Python list/dict

        discount_count = 0                                   # Counter for discounted items

        for product in products:
            if product['price'] >= min_price:                # Check price condition
                discount_amount = product['price'] * (discount_percent / 100)
                product['price'] = int(product['price'] - discount_amount)  # Apply discount
                discount_count += 1                          # Count updated item

        with open(output_file, 'w', encoding='utf-8') as f:  # Save updated JSON
            json.dump(products, f, indent=2, ensure_ascii=False)  # Unicode safe

        return discount_count                                # Return total discounted items

    except FileNotFoundError:
        print(f"File not found: {input_file}")               # Input file missing
        return 0
    except json.JSONDecodeError:
        print(f"Invalid JSON in file: {input_file}")         # JSON structure broken
        return 0


# Test the function
if __name__ == "__main__":                                   # Main check
    count = apply_discount(
        'data/products.json',
        10000,                                               # min price
        10,                                                  # 10% discount
        'data/discounted_products.json'
    )
    print(f"Discount applied to {count} products")           # Show result
```

```python
from pathlib import Path
import shutil

def copy_files_by_extension(source_dir, target_dir, extension):
    """
    Copy files with a specific extension from source directory to target directory.
    """
    source_path = Path(source_dir)                      # Source folder path
    target_path = Path(target_dir)                      # Target folder path

    target_path.mkdir(parents=True, exist_ok=True)      # Target folder မရှိရင် create လုပ်

    pattern = f"*.{extension}"                          # Extension filter (*.txt, *.pdf...)
    files_to_copy = list(source_path.glob(pattern))     # Matching files list

    copied_files = []

    for file_path in files_to_copy:                     # Loop through matched files
        target_file = target_path / file_path.name      # Destination file path
        shutil.copy(file_path, target_file)             # Copy file
        copied_files.append(file_path.name)             # Track copied files

    return copied_files                                 # Return copied filename list


# Example usage
if __name__ == "__main__":                              # Main run check
    copied = copy_files_by_extension("data", "backup/txt_files", "txt")
    print(f"Copied {len(copied)} files: {copied}")      # Show number of copied files
```

```python
import re   # regex library

def count_log_levels(log_file):
    """
    Count log entries by level (INFO, ERROR, WARNING, etc.)
    """
    level_count = {}                      # Level별 count သိမ်းမယ့် dict

    try:
        with open(log_file, 'r', encoding='utf-8') as file:   # Log file open
            for line in file:
                match = re.search(r'\] (\w+):', line)         # ] LEVEL: pattern capture
                if match:
                    level = match.group(1)                    # INFO / ERROR / WARNING ...
                    level_count[level] = level_count.get(level, 0) + 1
                    # Existing level count + 1

    except FileNotFoundError:
        print(f"Log file not found: {log_file}")

    return level_count                                        # Dict return


def find_error_entries(log_file):
    """
    Collect all ERROR lines from log file.
    """
    error_entries = []                                        # ERROR lines list

    try:
        with open(log_file, 'r', encoding='utf-8') as file:
            for line in file:
                if re.search(r'\] ERROR:', line):             # ERROR entry detection
                    error_entries.append(line.strip())        # Clean line add

    except FileNotFoundError:
        print(f"Log file not found: {log_file}")

    return error_entries                                      # List return


# Example usage
if __name__ == "__main__":                                    # Main run check

    counts = count_log_levels('data/application.log')         # Level count
    print("Log level counts:")
    for level, count in counts.items():
        print(f"  {level}: {count}")

    errors = find_error_entries('data/application.log')       # ERROR entries
    print(f"\nERROR entries found: {len(errors)}")
    for error in errors:
        print(f"  {error}")
```

```python
import csv      # CSV ဖိုင်ဖတ်/ရေးဖို့
import json     # JSON ဖိုင်ဖတ်/ရေးဖို့


def convert_value(value):
    """Convert string to correct data type."""
    
    # boolean ပြောင်းခြင်း
    if value.lower() == 'true':
        return True
    if value.lower() == 'false':
        return False

    # integer ပြောင်းခြင်း
    try:
        return int(value)
    except:
        pass

    # float ပြောင်းခြင်း
    try:
        return float(value)
    except:
        pass

    # အခြားအရာ—string အဖြစ်ထား
    return value


def csv_to_json(csv_file, json_file):
    """Read CSV → convert values → save JSON."""
    
    data = []    # converted row တွေကို သိမ်းမယ့် list

    # CSV ဖိုင်ဖွင့်
    with open(csv_file, 'r', encoding='utf-8') as f:
        reader = csv.DictReader(f)      # header row ကို key အဖြစ်သုံးတဲ့ reader

        for row in reader:              # CSV row တစ်ကြောင်းချင်း loop
            converted_row = {           # row dictionary ရဲ့ value တစ်ခုချင်း convert
                key: convert_value(value)
                for key, value in row.items()
            }
            data.append(converted_row)  # ပြောင်းပြီး row ကို list ထဲထည့်

    # JSON အဖြစ်သိမ်း (Unicode OK)
    with open(json_file, 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=2, ensure_ascii=False)

    return data                         # Converted data ပြန်ပေး


def test_convert_value():
    """Test conversion results."""
    print("Testing convert_value():")

    tests = [
        ("50000", 50000),        # int ဖြစ်သင့်
        ("4.5", 4.5),            # float ဖြစ်သင့်
        ("true", True),          # boolean True
        ("false", False),        # boolean False
        ("Ноутбук", "Ноутбук"),  # string တစ်ခြား
    ]

    # value convert result တွေ ပြသခြင်း
    for input_val, expected in tests:
        result = convert_value(input_val)
        status = "✓" if result == expected else "✗"    
        print(f"  {status} {input_val} -> {result}")


def test_csv_to_json():
    """Test the CSV-to-JSON conversion."""
    
    print("\nTesting csv_to_json():")

    try:
        # CSV ကို JSON ပြောင်း
        data = csv_to_json('data/mixed_data.csv', 'data/converted_data.json')

        first_row = data[0]     # type စစ်ဖို့ first row ယူ

        # type မှန်မမှန်စစ်ရမယ့်နေရာများ
        print("  Converted types:")
        print("   price:", type(first_row['price']))          # int ဖြစ်သင့်
        print("   rating:", type(first_row['rating']))        # float ဖြစ်သင့်
        print("   in_stock:", type(first_row['in_stock']))    # bool ဖြစ်သင့်
        print("   product:", type(first_row['product']))      # str ဖြစ်သင့်

        print(f"  Rows converted: {len(data)}")               # အကြောင်းအရေအတွက်
        print("  JSON saved successfully.")                   # JSON save success

    except Exception as e:
        print("  Error:", e)                                  # error handling


# Main entry point
if __name__ == "__main__":                                    # Python main run check
    test_convert_value()                                      # function test 1
    test_csv_to_json()                                        # function test 2
```
