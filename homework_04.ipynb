```python
def word_counter(file_path):
    """
    Counts lines, words, and characters (without spaces) in a text file.
    
    Args:
        file_path (str): Path to the text file
        
    Returns:
        dict: Dictionary with keys 'lines', 'words', 'chars'
    """
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            lines = file.readlines()
            
            # Initialize counters
            line_count = len(lines)
            word_count = 0
            char_count = 0
            
            # Process each line
            for line in lines:
                # Count words (split by whitespace)
                words = line.split()
                word_count += len(words)
                
                # Count characters without spaces
                char_count += len(line.replace(' ', ''))
            
            return {
                'lines': line_count,
                'words': word_count,
                'chars': char_count
            }
            
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        return None
    except Exception as e:
        print(f"Error reading file: {e}")
        return None

# Example usage with your sample file
result = word_counter('data/sample_text.txt')
if result:
    print(result)
```

```python
from pathlib import Path

def save_list_to_file(string_list, file_path):
    """
    Saves a list of strings to a text file, each element on a separate line.
    
    Args:
        string_list (list): List of strings to save
        file_path (str): Path where the file will be saved
        
    Returns:
        bool: True if successful, False otherwise
    """
    try:
        # Convert to Path object
        path = Path(file_path)
        
        # Create parent directories if they don't exist
        path.parent.mkdir(parents=True, exist_ok=True)
        
        # Write to file with UTF-8 encoding
        with open(path, 'w', encoding='utf-8') as file:
            for string in string_list:
                file.write(string + '\n')  # Add newline after each element
        
        print(f"Successfully saved {len(string_list)} lines to: {file_path}")
        return True
        
    except Exception as e:
        print(f"Error saving file: {e}")
        return False

# Example usage
if __name__ == "__main__":
    # Sample data
    my_list = [
        "Hello, World!",
        "Python programming",
        "File handling example",
        "နေကောင်းလား?",  # Burmese text
        "最后一行"  # Chinese text
    ]
    
    # Save to file (creates directories if needed)
    success = save_list_to_file(my_list, "output/my_files/data.txt")
    
    if success:
        print("File saved successfully!")
```

```python
def search_in_file(file_path, search_word):
    """
    Searches for lines containing a word (case-insensitive) in a text file.
    
    Args:
        file_path (str): Path to the text file
        search_word (str): Word to search for
        
    Returns:
        list: List of lines containing the search word
    """
    try:
        found_lines = []
        
        with open(file_path, 'r', encoding='utf-8') as file:
            for line in file:
                # Case-insensitive search
                if search_word.lower() in line.lower():
                    found_lines.append(line.strip())
        
        return found_lines
        
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        return []
    except Exception as e:
        print(f"Error reading file: {e}")
        return []

# Example usage with your sample file
if __name__ == "__main__":
    # Test with different search words
    search_words = ["python", "данных", "синтаксис", "programming"]
    
    for word in search_words:
        print(f"\nSearching for '{word}':")
        results = search_in_file('data/sample_text.txt', word)
        
        if results:
            print(f"Found {len(results)} line(s):")
            for i, line in enumerate(results, 1):
                print(f"{i}. {line}")
        else:
            print("No matches found.")
```

```python
import csv

def read_csv_to_dict(file_path):
    """
    Reads a CSV file and returns a list of dictionaries.
    
    Args:
        file_path (str): Path to the CSV file
        
    Returns:
        list: List of dictionaries where keys are column headers
    """
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            # Use DictReader to automatically use first line as headers
            csv_reader = csv.DictReader(file)
            
            # Convert to list of dictionaries
            data = list(csv_reader)
            
            return data
            
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        return []
    except Exception as e:
        print(f"Error reading CSV file: {e}")
        return []

# Example usage with your CSV file
if __name__ == "__main__":
    # Read the CSV file
    users = read_csv_to_dict('data/users.csv')
    
    # Display the results
    print("CSV Data as List of Dictionaries:")
    print("-" * 40)
    
    for i, user in enumerate(users, 1):
        print(f"Record {i}:")
        for key, value in user.items():
            print(f"  {key}: {value}")
        print()
```

```python
from pathlib import Path

def merge_files(file_paths, output_file):
    """
    Merges multiple text files into one with headers.
    
    Args:
        file_paths (list): List of file paths to merge
        output_file (str): Path for the output file
        
    Returns:
        int: Count of successfully merged files
    """
    merged_count = 0
    
    # Create output directory if it doesn't exist
    Path(output_file).parent.mkdir(parents=True, exist_ok=True)
    
    with open(output_file, 'w', encoding='utf-8') as outfile:
        for i, file_path in enumerate(file_paths, 1):
            try:
                with open(file_path, 'r', encoding='utf-8') as infile:
                    content = infile.read().strip()
                    
                    # Add header before each file content
                    outfile.write(f"=== File {i}: {Path(file_path).name} ===\n")
                    outfile.write(content)
                    outfile.write("\n\n")
                    
                    merged_count += 1
                    
            except FileNotFoundError:
                continue  # Skip non-existent files
    
    return merged_count

# Example usage
files = ['data/file1.txt', 'data/file2.txt', 'data/file3.txt']
result = merge_files(files, 'data/merged/combined.txt')
print(f"Merged {result} files")
```

```python
import csv

def filter_csv(file_path, output_path, filter_func):
    """
    Filters CSV rows by condition and saves to new file.
    
    Args:
        file_path (str): Input CSV file path
        output_path (str): Output CSV file path  
        filter_func (function): Lambda function for filtering
        
    Returns:
        int: Count of filtered rows
    """
    filtered_count = 0
    
    with open(file_path, 'r', encoding='utf-8') as infile, \
         open(output_path, 'w', encoding='utf-8', newline='') as outfile:
        
        reader = csv.DictReader(infile)
        writer = csv.DictWriter(outfile, fieldnames=reader.fieldnames)
        
        writer.writeheader()
        
        for row in reader:
            if filter_func(row):
                writer.writerow(row)
                filtered_count += 1
    
    return filtered_count

# Test functions
def test_filter_csv():
    """Test the filter_csv function with different conditions"""
    
    # Test 1: Filter by city (Москва)
    count1 = filter_csv('data/employees.csv', 'data/moscow_employees.csv', 
                       lambda row: row['city'] == 'Москва')
    print(f"Москва employees: {count1}")
    
    # Test 2: Filter by age (>30)
    count2 = filter_csv('data/employees.csv', 'data/senior_employees.csv',
                       lambda row: int(row['age']) > 30)
    print(f"Employees over 30: {count2}")
    
    # Test 3: Filter by salary (>=90000)
    count3 = filter_csv('data/employees.csv', 'data/high_salary.csv',
                       lambda row: int(row['salary']) >= 90000)
    print(f"High salary employees: {count3}")

# Run tests
if __name__ == "__main__":
    test_filter_csv()
```

```python
import json

def apply_discount(input_file, min_price, discount_percent, output_file):
    """
    Apply discount to products above certain price and save to new JSON file.
    
    Args:
        input_file (str): Input JSON file path
        min_price (int): Minimum price to apply discount
        discount_percent (float): Discount percentage (0-100)
        output_file (str): Output JSON file path
        
    Returns:
        int: Count of products that received discount
    """
    try:
        with open(input_file, 'r', encoding='utf-8') as f:
            products = json.load(f)
        
        discount_count = 0
        
        for product in products:
            if product['price'] >= min_price:
                # Apply discount
                discount_amount = product['price'] * (discount_percent / 100)
                product['price'] = int(product['price'] - discount_amount)
                discount_count += 1
        
        with open(output_file, 'w', encoding='utf-8') as f:
            json.dump(products, f, indent=2, ensure_ascii=False)
        
        return discount_count
        
    except FileNotFoundError:
        print(f"File not found: {input_file}")
        return 0
    except json.JSONDecodeError:
        print(f"Invalid JSON in file: {input_file}")
        return 0

# Test the function
if __name__ == "__main__":
    count = apply_discount('data/products.json', 10000, 10, 'data/discounted_products.json')
    print(f"Discount applied to {count} products")
```

```python
from pathlib import Path
import shutil

def copy_files_by_extension(source_dir, target_dir, extension):
    """
    Copy files with specific extension to target directory.
    
    Args:
        source_dir (str): Source directory path
        target_dir (str): Target directory path  
        extension (str): File extension to copy (e.g., 'txt', 'pdf')
        
    Returns:
        list: List of copied filenames
    """
    source_path = Path(source_dir)
    target_path = Path(target_dir)
    
    # Create target directory if needed
    target_path.mkdir(parents=True, exist_ok=True)
    
    # Find files with specified extension
    pattern = f"*.{extension}"
    files_to_copy = list(source_path.glob(pattern))
    
    copied_files = []
    
    # Copy each file
    for file_path in files_to_copy:
        target_file = target_path / file_path.name
        shutil.copy(file_path, target_file)
        copied_files.append(file_path.name)
    
    return copied_files

# Example usage
if __name__ == "__main__":
    copied = copy_files_by_extension("data", "backup/txt_files", "txt")
    print(f"Copied {len(copied)} files: {copied}")
```

```python
import re

def count_log_levels(log_file):
    """
    Count log entries by level (INFO, ERROR, WARNING, etc.)
    
    Args:
        log_file (str): Path to log file
        
    Returns:
        dict: Count of entries by log level
    """
    level_count = {}
    
    try:
        with open(log_file, 'r', encoding='utf-8') as file:
            for line in file:
                # Use regex to extract log level
                match = re.search(r'\] (\w+):', line)
                if match:
                    level = match.group(1)
                    level_count[level] = level_count.get(level, 0) + 1
    
    except FileNotFoundError:
        print(f"Log file not found: {log_file}")
    
    return level_count

def find_error_entries(log_file):
    """
    Find all ERROR entries in log file
    
    Args:
        log_file (str): Path to log file
        
    Returns:
        list: All ERROR log entries
    """
    error_entries = []
    
    try:
        with open(log_file, 'r', encoding='utf-8') as file:
            for line in file:
                # Use regex to find ERROR entries
                if re.search(r'\] ERROR:', line):
                    error_entries.append(line.strip())
    
    except FileNotFoundError:
        print(f"Log file not found: {log_file}")
    
    return error_entries

# Example usage
if __name__ == "__main__":
    # Count log levels
    counts = count_log_levels('data/application.log')
    print("Log level counts:")
    for level, count in counts.items():
        print(f"  {level}: {count}")
    
    # Find ERROR entries
    errors = find_error_entries('data/application.log')
    print(f"\nERROR entries found: {len(errors)}")
    for error in errors:
        print(f"  {error}")
```

```python
import csv
import json

def convert_value(value):
    """
    Convert string to appropriate type (int, float, bool, or keep as string)
    """
    # Try converting to boolean
    if value.lower() == 'true':
        return True
    elif value.lower() == 'false':
        return False
    
    # Try converting to integer
    try:
        return int(value)
    except ValueError:
        pass
    
    # Try converting to float
    try:
        return float(value)
    except ValueError:
        pass
    
    # Return as string if no conversion works
    return value

def csv_to_json(csv_file, json_file):
    """
    Read CSV and convert to JSON with proper type conversion
    """
    data = []
    
    with open(csv_file, 'r', encoding='utf-8') as f:
        reader = csv.DictReader(f)
        
        for row in reader:
            # Convert all values to appropriate types
            converted_row = {key: convert_value(value) for key, value in row.items()}
            data.append(converted_row)
    
    # Save as JSON
    with open(json_file, 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=2, ensure_ascii=False)
    
    return data

# Test functions
def test_convert_value():
    """Test the type conversion function"""
    print("Testing convert_value():")
    
    tests = [
        ("50000", 50000),      # string to int
        ("4.5", 4.5),          # string to float  
        ("true", True),        # string to boolean
        ("false", False),      # string to boolean
        ("Ноутбук", "Ноутбук") # string remains string
    ]
    
    for input_val, expected in tests:
        result = convert_value(input_val)
        status = "✓" if result == expected else "✗"
        print(f"  {status} convert_value('{input_val}') = {result} (expected: {expected})")

def test_csv_to_json():
    """Test the CSV to JSON conversion"""
    print("\nTesting csv_to_json():")
    
    try:
        data = csv_to_json('data/mixed_data.csv', 'data/converted_data.json')
        
        # Check if types are correctly converted
        first_row = data[0]
        
        type_checks = [
            (type(first_row['price']), int, 'price should be int'),
            (type(first_row['rating']), float, 'rating should be float'),
            (type(first_row['in_stock']), bool, 'in_stock should be bool'),
            (type(first_row['product']), str, 'product should be str')
        ]
        
        print("  Type conversion checks:")
        for actual_type, expected_type, message in type_checks:
            status = "✓" if actual_type == expected_type else "✗"
            print(f"    {status} {message}: {actual_type}")
        
        print(f"  ✓ Converted {len(data)} rows")
        print(f"  ✓ JSON file saved: data/converted_data.json")
        
    except Exception as e:
        print(f"  ✗ Error: {e}")

# Run tests
if __name__ == "__main__":
    test_convert_value()
    test_csv_to_json()
```
